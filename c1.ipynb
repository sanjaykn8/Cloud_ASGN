```python
N = 1000000
sc = pyspark.SparkContext("local[*]", "spark-euler")
sc.setLogLevel("ERROR")
print(f"Spark {sc.version} | default parallelism = {sc.defaultParallelism} | N = {N:,}\n")
```

```python
t0 = time.time()
basel = (sc.parallelize(range(1, N + 1), sc.defaultParallelism * 4)
           .map(lambda n: 1.0 / (n * n))
           .reduce(lambda a, b: a + b))
print("Basel sum  sum(1/n^2) =", basel)
print("pi^2 / 6              =", math.pi ** 2 / 6)
print(f"abs error             = {abs(basel - math.pi ** 2 / 6):.3e}   ({time.time() - t0:.1f}s)\n")
```

```python
e_approx = (sc.parallelize(range(0, 30))
              .map(lambda k: 1.0 / math.factorial(k))
              .sum())
print("Euler's number  sum(1/k!) =", e_approx)
print("math.e                    =", math.e)
print(f"abs error                 = {abs(e_approx - math.e):.3e}")

sc.stop()
```
